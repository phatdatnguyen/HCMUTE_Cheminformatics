#### 1.1.2.5. If Statement

An if statement is written by using the `if` keyword. The body of an if statement is indented. For example:

```python
a = 33
b = 200
if b > a:
    print("b is greater than a")
```

The `elif` keyword is Python's way of saying "if the previous conditions were not true, then try this condition".
The `else` keyword catches anything which isn't caught by the preceding conditions.

```python
a = 200
b = 33
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
else:
    print("a is greater than b")
```

#### 1.1.2.6. While Loop

A `while` loop repeats its indented body while a condition is true. Ensure that the condition can eventually become false; here `i += 1` advances the loop.

```python
i = 1
while i < 6:
    print(i)
    i += 1
```

With the `break` statement we can stop the loop even if the while condition is `True`:

```python
i = 1
while i < 6:
    print(i)
    if i == 3:
        break
    i += 1 
```

With the `continue` statement we can stop the current iteration, and continue with the next:

```python
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
```

#### 1.1.2.7. For Loop

A `for` loop visits each item in an **iterable**, such as a list, tuple, string, dictionary, set, or `range`. Dictionaries yield keys by default; sets have no guaranteed traversal order. Indent the loop body with four spaces.

```python
# Loop through a list
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(x)
```

```python
# Loop through a string
for x in "banana":
    print(x)
```

With the `break` statement we can stop the loop before it has looped through all the items:

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(x)
    if x == "banana":
        break
```

With the continue statement we can stop the current iteration, and continue with the next:

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    if x == "banana":
        continue
    print(x)
```

#### 1.1.2.8. Function

Define a reusable function with `def`, choose descriptive parameter names, and use `return` to provide a result to the caller. `print` only displays text.

This example computes the length of a two-component vector, $\sqrt{x^2+y^2}$. Inputs must use the same unit; the result has that unit. `math.hypot` implements this calculation robustly.

```python
def vector_length(x, y):
    """Return a 2D vector length; x and y must use the same units."""
    return math.hypot(x, y)
```

Now you can test this function

```python
print(vector_length(3, 4))
print(vector_length(6, 8))
assert math.isclose(vector_length(3, 4), 5.0)
```

### A small chemical calculation: concentration

An **amount of substance** $n$ counts entities in moles. A **molar mass** $M$ converts grams to moles; an **amount concentration** $c$ tells us how many moles are present per liter of solution. Here “0.1 M” means “0.1 mol/L”; distinguish this unit abbreviation from the symbol $M$ for molar mass.

$$n=\frac{m}{M},\qquad c=\frac{n}{V}=\frac{m}{MV}.$$

| Input | Meaning | Unit in our function |
| --- | --- | --- |
| `mass_g` | Mass of solute | g |
| `molar_mass_g_mol` | Mass per mole | g/mol |
| `volume_l` | **Final solution volume** | L |

Assume pure NaCl, the stated molar mass, complete dissolution, and preparation to the stated final volume. These are calculated preparation values, not measured concentrations. Checking finite numbers and positive divisors protects the arithmetic; it cannot verify purity or experimental technique.

**Predict:** dissolving twice the mass and making up to the same final volume should do what to $c$?

```python
def molar_concentration(mass_g, molar_mass_g_mol, volume_l):
    """Return concentration in mol/L for finite, physically valid inputs."""
    if not all(math.isfinite(value) for value in (mass_g, molar_mass_g_mol, volume_l)):
        raise ValueError("Inputs must be finite numbers.")
    if mass_g < 0 or molar_mass_g_mol <= 0 or volume_l <= 0:
        raise ValueError("Mass must be non-negative; molar mass and volume must be positive.")
    return mass_g / molar_mass_g_mol / volume_l

# Illustrative NaCl preparation, using a rounded molar mass.
concentration_mol_l = molar_concentration(0.5844, 58.44, 0.1000)
print(f"Concentration: {concentration_mol_l:.4f} mol/L")
assert math.isclose(concentration_mol_l, 0.1000)
```

### Research application: plan a small concentration series

Suppose a solution-screening experiment needs **10.0 mL each** of 1, 2, 5, and 10 mmol/L NaCl, prepared from the 0.100 mol/L stock calculated above. A **stock** is a more concentrated starting solution; an **aliquot** is the portion transferred from it. Here the concentrations are design targets, not experimental observations.

Conservation of the transferred amount gives $c_{\rm stock}V_{\rm stock}=c_{\rm target}V_{\rm final}$. First convert mmol/L to mol/L; then solve for the aliquot volume. Make each aliquot up to the **final solution volume**, rather than assuming that separately measured solution volumes add exactly.

**Predict:** which target requires the largest aliquot? Would doubling every final volume change the required stock concentration or the aliquot volumes?

```python
stock_mol_L = concentration_mol_l
final_volume_mL = 10.0
targets_mmol_L = [1.0, 2.0, 5.0, 10.0]
dilution_plan = []
for target_mmol_L in targets_mmol_L:
    target_mol_L = target_mmol_L / 1000.0
    aliquot_mL = target_mol_L * final_volume_mL / stock_mol_L
    if not 0 < aliquot_mL <= final_volume_mL:
        raise ValueError("This target cannot be made by diluting the stock.")
    dilution_plan.append({"target_mmol_L": target_mmol_L,
                          "aliquot_uL": aliquot_mL * 1000.0})
    recovered = stock_mol_L * aliquot_mL / final_volume_mL
    assert math.isclose(recovered, target_mol_L)
    print(f"{target_mmol_L:4.1f} mmol/L: transfer {aliquot_mL*1000:6.1f} microliters; "
          f"make up to {final_volume_mL:.1f} mL")
```

The next cell supplies plotting instructions; focus on reading the figure. Part 2 explains `fig`, `ax`, and arrays. The arrows show how units connect the calculation, and the bars make the required transfers easy to compare.

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 1, figsize=(8, 5.6), height_ratios=[1, 2],
                         constrained_layout=True)
axes[0].set(xlim=(0, 1), ylim=(0, 1))
axes[0].axis("off")
for position, label in [(0.14, "0.5844 g NaCl"), (0.5, "0.0100 mol"),
                        (0.86, "0.100 mol/L")]:
    axes[0].text(position, 0.55, label, ha="center", va="center",
                 bbox={"boxstyle": "round,pad=0.5", "facecolor": "#dbeafe"})
for start, end, label in [(0.25, 0.39, "divide by\n58.44 g/mol"),
                           (0.61, 0.75, "divide by\n0.100 L")]:
    axes[0].annotate("", xy=(end, 0.55), xytext=(start, 0.55),
                     arrowprops={"arrowstyle": "->"})
    axes[0].text((start+end)/2, 0.1, label, ha="center", fontsize=9)
labels = [f"{row['target_mmol_L']:g} mmol/L" for row in dilution_plan]
aliquots = [row["aliquot_uL"] for row in dilution_plan]
bars = axes[1].barh(labels, aliquots, color="#2563eb")
axes[1].bar_label(bars, labels=[f"{value:g}" for value in aliquots], padding=4)
axes[1].set(xlabel="Stock aliquot (microliters)", xlim=(0, 1200),
            title="Calculated plan: each final solution is 10.0 mL")
plt.show()
```
